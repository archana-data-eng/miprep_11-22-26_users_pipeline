```python
import pandas as pd
import sqlite3
from google.colab import files

df =pd.read_excel("miprep_11-22-26_users_xlsx", header=1)
df['FIRST NAME'] = df['FIRST NAME'].astype(str).str.strip().str.title()
df['LAST NAME'] = df['LAST NAME'].astype(str).str.strip().str.title()
df['TARGET_ROLE'] = df['TARGET_ROLE'].astype(str).str.strip().str.title()
df['SESSION_STATUS'] = df['SESSION_STATUS'].astype(str).str.strip().str.title()

df['EMAIL'] = df['EMAIL'].astype(str).str.replace(" ", "").str.lower()

df['MOCK_SCORE_PERCENT'] = pd.to_numeric(df['MOCK_SCORE_PERCENT'], errors='coerce')

print(df['SESSION_STATUS'].value_counts())

print("AVERAGE SCORE:", round(df['MOCK_SCORE_PERCENT'].mean(), 2))

df.to_excel("cleaned_miprep_11-22-26_users_xlsx", index=False)

conn = sqlite3.connect("miprep_clean.db")
df.to_sql("clean_users", conn, if_exists="replace", index=False)
conn.close()

files.download("cleaned_miprep_11-22-26_users_xlsx")
df.head(5)
```

```python
import pandas as pd
import sqlite3
from google.colab import files

# Fix: Load the correct filename from disk
df = pd.read_excel("miprep_11-22-26_users_xlsx.xlsx", header=1)

# Fix name mapping so FIRST NAME handles FIRST NAME and LAST NAME handles LAST NAME
df['FIRST NAME'] = df['FIRST NAME'].astype(str).str.strip().str.title()
df['LAST NAME'] = df['LAST NAME'].astype(str).str.strip().str.title()
df['TARGET_ROLE'] = df['TARGET_ROLE'].astype(str).str.strip().str.title()
df['SESSION_STATUS'] = df['SESSION_STATUS'].astype(str).str.strip().str.title()

df['EMAIL'] = df['EMAIL'].astype(str).str.replace(" ", "").str.lower()

df['MOCK_SCORE_PERCENT'] = pd.to_numeric(df['MOCK_SCORE_PERCENT'], errors='coerce')

print(df['SESSION_STATUS'].value_counts())
print("AVERAGE SCORE:", round(df['MOCK_SCORE_PERCENT'].mean(), 2))

# Fix: Added the correct .xlsx extension so Pandas knows which writer to use
cleaned_filename = "cleaned_miprep_11-22-26_users.xlsx"
df.to_excel(cleaned_filename, index=False)

conn = sqlite3.connect("miprep_clean.db")
df.to_sql("clean_users", conn, if_exists="replace", index=False)
conn.close()

files.download(cleaned_filename)
df.head(5)
```
